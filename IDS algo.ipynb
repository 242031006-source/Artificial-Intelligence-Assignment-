```python
import networkx as nx
import matplotlib.pyplot as plt
import time
def dls(node,goal,depth,visited):
  if node==goal:
    print(node,end=" ")
    visited.add(node)
    show_graph(visited)
    return True
  if depth==0:
    return False
  visited.add(node)
  print(node,end=" ")
  show_graph(visited)
  for neighbour in graph[node]:
    if neighbour not in visited:
      if dls(neighbour,goal,depth-1,visited):
        return True
  return False
def ids(start,goal):
  for depth in range(0,max_depth+1):
    print("\ndepth",depth,":",end=" ")
    visited=set()
    if dls(start,goal,depth,visited):
      print("\ngoal found")
      return
  print("\ngoal not found",max_depth)
graph={
     1:[2,3],
     2:[1,4,5],
     3:[1,6,7],
     4:[2],
     5:[2,8],
     6:[3,8],
     7:[3],
     8:[5,6]
}
G=nx.Graph()
for node in graph:
  for neighbour in graph[node]:
    G.add_edge(node,neighbour)
pos = {
    1: (0, 3),
    2: (-2, 2),
    3: (2, 2),
    4: (-3, 1),
    5: (-1, 1),
    6: (1, 1),
    7: (3, 1),
    8: (0, 0)
}
def show_graph(visited):
  colors=[]
  for n in G.nodes():
    if n in visited:
      colors.append("lightgreen")
    else:
      colors.append("lightblue")
  plt.figure(figsize=(10,6))
  nx.draw(
      G,pos,with_labels=True,
      node_color=colors,
      node_size=1400,
      font_size=14,
      font_weight="bold",
      edge_color="gray",
      width=2
  )
  plt.title("IDS Graph Visualization",fontsize=16)
  plt.axis("off")
  plt.show()
  time.sleep(1)
start=int(input("Enter Starting node:"))
goal=int(input("Enter goal node:"))
max_depth=int(input("Enter max depth:"))
ids(start,goal)
```
