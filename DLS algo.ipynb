```python
import networkx as nx
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output
def dls(graph, start, limit, visited=None, order=None, depth=0):
    if visited is None:
        visited = set()
    if order is None:
        order = []

    visited.add(start)
    order.append((start, depth))
    clear_output(wait=True)
    print("Current Node:", start)
    print("Current Depth:", depth)
    print("DLS Traversal:", order)
    colors = []
    for n in G.nodes():
        if n in visited:
            colors.append("lightgreen")
        else:
            colors.append("lightblue")

    plt.figure(figsize=(10, 6))
    nx.draw(
        G, pos,
        with_labels=True,
        node_color=colors,
        node_size=1400,
        font_size=14,
        font_weight="bold",
        edge_color="gray",
        width=2
    )

    plt.title("DLS Graph Visualization", fontsize=16)
    plt.axis("off")
    plt.show()
    time.sleep(1)
    if depth < limit:
        for neighbour in graph[start]:
            if neighbour not in visited:
                dls(graph, neighbour, limit, visited, order, depth + 1)

    return order
graph = {
    1: [2, 3],
    2: [1, 4, 5],
    3: [1, 6, 7],
    4: [2],
    5: [2, 8],
    6: [3, 8],
    7: [3],
    8: [5, 6]
}
G = nx.Graph()

for node in graph:
    for neighbour in graph[node]:
        G.add_edge(node, neighbour)
pos = {
    1: (0, 3),
    2: (-2, 2),
    3: (2, 2),
    4: (-3, 1),
    5: (-1, 1),
    6: (1, 1),
    7: (3, 1),
    8: (0, 0)
}
start_node = int(input("Enter the starting node (1-8): "))
depth_limit = int(input("Enter the depth limit: "))
result = dls(graph, start_node, depth_limit)
print("DLS Traversal:", result)
```
